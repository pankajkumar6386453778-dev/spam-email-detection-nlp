```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text = 'I Love this movie! It was amazing and exiciting.'
```

```python
print(text)
```

## Lowercase

```python
text = text.lower()
print(text)
```

## removing punctuation marks

```python
import string
text = text.translate(str.maketrans('','',string.punctuation))
print(text)
```

## Tokenization

```python
token = word_tokenize(text)
print(token)
```

## stop words removal

```python
stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
filtered_word = []
for word in token:
    if word not in stop_words:
        filtered_word.append(word)
    print(filtered_word)
```

## Lemmatization

```python
nltk.download('wordnet')
lemmatizer = nltk.WordNetLemmatizer()
filtered_word = [
    lemmatizer.lemmatize(word)
    for word in filtered_word
]
print(filtered_word)
```

```python
## imdb rating distribution
plt.hist(df['IMDB_Rating'] , bins = 10)
plt.title("Distribution of imdb rating")
plt.xlabel('imdb rating')
plt.ylabel('number of movies')
plt.show()
```

```python
## top 10 movie generes
df['Genre'] = df['Genre'].str.split(',').str[0]
genre_count = df['Genre'].value_counts().head(10)
```

```python
genre_count.plot(kind='bar')
plt.title('top 10 movies genres')
plt.xlabel('genre')
plt.ylabel('number of movies')
plt.show()
```

```python
## average rating of genres
average_rating = df.groupby('Genre')['IMDB_Rating'].mean()
```

```python
average_rating = average_rating.sort_values(ascending=False).head(10)
average_rating.plot(kind='bar')
plt.title('average rating of genre')
plt.xlabel('genre')
plt.ylabel('average rating')
plt.show()
```
